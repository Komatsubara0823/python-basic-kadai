```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
          print(self.name, self.age,"adult")
        else:
          print(self.name, self.age,"child")

# インスタンス化する
human = [
    Human("侍太郎", 36),
    Human("侍花子", 20),
    Human("侍次郎", 16)
]
for person in human:
    person.check_adult()
```
