```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def printinfo(self):
        print(self.name, self.age)

    def check_adult(self):
        if self.age > 20:
          print("adult")
        else:
          print("child")

# インスタンス化する
human = [
    Human("侍太郎", 36),
    Human("侍花子", 26),
    Human("侍次郎", 16)
]
for person in human:
    person.printinfo()
    person.check_adult()
```
